DATA ANALYSIS ON CRUDE OIL USAGE PER YEAR

 # INTRODUCTION

Crude oil is a vital resource that powers industries, transportation, and economies worldwide. Analyzing its usage trends over the years provides valuable insights into supply-demand dynamics, production efficiency, and market fluctuations. This study aims to examine crude oil usage per year by leveraging historical data to identify key patterns, variations, and factors influencing oil consumption. By analyzing annual trends, we can assess fluctuations in production, identify peak and low consumption periods, and evaluate the impact of economic or geopolitical events on oil usage. The findings from this analysis can assist policymakers, investors, and industry stakeholders in making data-driven decisions for better resource management and future planning


# DESCRIPTION:

 The dataset comprises 1,392 records with four columns representing monthly crude oil data from various oil companies over different years. The "Month" column contains the names of months, indicating the period of data collection, while the "Year" column specifies the corresponding year. The "OIL COMPANIES" column lists the names of different oil companies involved in crude oil production or distribution. The "Quantity (000 Metric Tonnes)" column records the amount of crude oil handled by each company in a given month, measured in thousand metric tonnes. However, there are four missing values in the quantity column, which may require data cleaning before analysis. This dataset is valuable for examining yearly and monthly crude oil trends, comparing production across companies, and analyzing overall oil consumption patterns

# crude oil usage by months

```python

df_cleaned = df.dropna()

df_cleaned["Month"] = df_cleaned["Month"].astype(str)

monthly_usage = df_cleaned.groupby("Month")["Quantity (000 Metric Tonnes)"].sum().reset_index()

month_order = ["January", "February", "March", "April", "May", "June", 
               "July", "August", "September", "October", "November", "December"]
monthly_usage["Month"] = pd.Categorical(monthly_usage["Month"], categories=month_order, ordered=True)
monthly_usage = monthly_usage.sort_values("Month")

plt.figure(figsize=(10, 5))
sns.barplot(data=monthly_usage, x="Month", y="Quantity (000 Metric Tonnes)", palette="Blues_d")
plt.xticks(rotation=45)
plt.ylabel("Total Crude Oil Quantity (000 Metric Tonnes)")
plt.title("Crude Oil Consumption by Month")
plt.show()

monthly_usage
```

```python
from scipy.stats import ttest_ind
bpcl_data = df[df["OIL COMPANIES"] == "ONGC TOTAL"]["Quantity (000 Metric Tonnes)"]
hpcl_data = df[df["OIL COMPANIES"] == "IOCL TOTAL"]["Quantity (000 Metric Tonnes)"]
t_stat, p_value = ttest_ind(bpcl_data, hpcl_data, equal_var=False)
print(f"T-Statistic: {t_stat}")
print(f"P-Value: {p_value}")
if p_value < 0.05:
    print("Statistically significant difference in crude oil usage between ONGC and IOCL.")
else:
    print("No significant difference in crude oil usage between ONGC and IOCL.")
```

```python
from scipy.stats import ttest_ind
bpcl_data = df[df["OIL COMPANIES"] == "BPCL-TOTAL"]["Quantity (000 Metric Tonnes)"]
hpcl_data = df[df["OIL COMPANIES"] == "HPCL-TOTAL"]["Quantity (000 Metric Tonnes)"]
t_stat, p_value = ttest_ind(bpcl_data, hpcl_data, equal_var=False)
print(f"T-Statistic: {t_stat}")
print(f"P-Value: {p_value}")
if p_value < 0.05:
    print("Statistically significant difference in crude oil usage between BPCL-TOTAL and HPCL-TOTAL.")
else:
    print("No significant difference in crude oil usage between BPCL-TOTAL and HPCL-TOTAL.")
```

# WHICH COMPANY USES MORE CRUDE OIL IN LAST 5 YEARS

```python
df["Year"] = pd.to_numeric(df["Year"], errors="coerce")

latest_year = df["Year"].max()
df_last_5_years = df[df["Year"] >= latest_year - 4]

company_usage_5_years = df_last_5_years.groupby("OIL COMPANIES")["Quantity (000 Metric Tonnes)"].sum().reset_index()

top_company = company_usage_5_years.sort_values("Quantity (000 Metric Tonnes)", ascending=False).iloc[0]

plt.figure(figsize=(12, 6))
sns.barplot(
    data=company_usage_5_years.sort_values("Quantity (000 Metric Tonnes)", ascending=False).head(10),
    x="Quantity (000 Metric Tonnes)",
    y="OIL COMPANIES",
    palette="magma"
)
plt.xlabel("Total Crude Oil Consumption (000 Metric Tonnes)")
plt.ylabel("Oil Companies")
plt.title(f"Top 10 Companies by Crude Oil Consumption in Last 5 Years ({latest_year-4} - {latest_year})")
plt.grid(axis="x", linestyle="--", alpha=0.7)
plt.show()

print(f"\nThe company that used the most crude oil in the last 5 years is: **{top_company['OIL COMPANIES']}**")
print(f"Total crude oil consumption: **{top_company['Quantity (000 Metric Tonnes)']:.2f}** (000 Metric Tonnes)")
```

# WHICH COMPANY USES LOW CRUDE OIL IN LAST 5 YEARS

```python
df["Year"] = pd.to_numeric(df["Year"], errors="coerce")

latest_year = df["Year"].max()
df_last_5_years = df[df["Year"] >= latest_year - 4]

company_usage_5_years = df_last_5_years.groupby("OIL COMPANIES")["Quantity (000 Metric Tonnes)"].sum().reset_index()

lowest_company = company_usage_5_years.sort_values("Quantity (000 Metric Tonnes)", ascending=True).iloc[0]

plt.figure(figsize=(12, 6))
sns.barplot(
    data=company_usage_5_years.sort_values("Quantity (000 Metric Tonnes)", ascending=True).head(10),
    x="Quantity (000 Metric Tonnes)",
    y="OIL COMPANIES",
    palette="coolwarm"
)
plt.xlabel("Total Crude Oil Consumption (000 Metric Tonnes)")
plt.ylabel("Oil Companies")
plt.title(f"Bottom 10 Companies by Crude Oil Consumption in Last 5 Years ({latest_year-4} - {latest_year})")
plt.grid(axis="x", linestyle="--", alpha=0.7)
plt.show()

print(f"\nThe company that used the least crude oil in the last 5 years is: **{lowest_company['OIL COMPANIES']}**")
print(f"Total crude oil consumption: **{lowest_company['Quantity (000 Metric Tonnes)']:.2f}** (000 Metric Tonnes)")
```
